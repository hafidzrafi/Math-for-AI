```python
import numpy as np
import matplotlib.pyplot as plt
from main import vecs_visualize
```

```python
def norm_vec(vectors: list[np.ndarray]):
    vec_norm = []
    
    for vec in vectors:
        norm = np.linalg.norm(vec)
        vec_norm.append(vec/norm)
        
    return vec_norm
```

```python
utara = np.array([0, 5])
timur_laut = np.array([4, 4])
hampir_timur = np.array([5, 1])

vecs = norm_vec([utara, timur_laut, hampir_timur])

print(f"{[vecs[i] @ vecs[i+1] for i in range(len(vecs) - 1) ]}")

vecs_visualize([utara, timur_laut, hampir_timur], ["blue", "green", "red"])
```

# 🚀 MISSION 2: THE SIMILARITY SCANNER
### "Building the Engine of Recommendation Systems"

**TUJUAN:** Membuktikan secara matematis film mana yang paling cocok untuk User berdasarkan preferensi genre menggunakan operasi **Dot Product**.

---

### 📊 DATA INTELLIGENCE
Kita bekerja di **Ruang Fitur 3-Dimensi** dengan sumbu: `[Action, Romance, Comedy]`.

**1. TARGET (The User):**
* **👤 User Profile:** `[5, 1, 0]`  
    *(Sangat suka Action, sedikit Romance, benci Comedy)*

**2. KANDIDAT (The Movies):**
* **🎬 Film A (The Rambo):** `[4, 0, 1]`  
    *(Dominan Action)*
* **🎬 Film B (Romeo & Juliet):** `[0, 5, 1]`  
    *(Dominan Romance)*
* **🎬 Film C (Rush Hour):** `[2, 2, 5]`  
    *(Dominan Comedy)*

---

### 🧠 KONSEP MATEMATIKA
Kita menggunakan **Dot Product** sebagai pengukur kemiripan (*Similarity Metric*).
$$\text{Score} = \mathbf{u} \cdot \mathbf{v} = \sum_{i=1}^{n} u_i v_i$$

* **Skor Tinggi:** Vektor searah (Mirip/Cocok).
* **Skor Rendah/Nol:** Vektor tegak lurus (Tidak Cocok).

---

### 📝 INSTRUKSI TUGAS
1.  **Inisialisasi:** Buat NumPy array untuk `user`, `film_a`, `film_b`, dan `film_c`.
2.  **Komputasi:** Hitung Dot Product antara user dengan setiap film menggunakan operator `@` atau `np.dot()`.
3.  **Analisis:** Bandingkan ketiga skor tersebut.
4.  **Visualisasi:** Buat **Bar Chart** sederhana untuk menampilkan skor rekomendasi.

> **Constraint:** DILARANG menggunakan loop `for` untuk perkalian. Gunakan fitur *vectorization* NumPy.

```python
user = np.array([3, 2, 0])

film_a = np.array([5,0,5])
film_b = np.array([1,5,2])
film_c = np.array([5,3,0])
```

```python
skor_a = film_a @ user
skor_b = film_b @ user
skor_c = film_c @ user
```

```python
labels = ["Film a", "Film b", "Film c"]
scores = [skor_a, skor_b, skor_c]
colors = ["red", "green", "blue"]

plt.style.use("dark_background")

plt.figure(figsize=(8,5))
plt.title("Recommendation Systems")
plt.ylabel("Score")
plt.grid(axis='y', linestyle='--', alpha=0.3)

bars = plt.bar(labels, scores, color=colors)

plt.bar_label(bars)
```

```python

# 1. MATRIKS DATA (X) - Tumpukan 3 Film
# Bayangkan ini tabel Excel: Baris = Film, Kolom = Genre
X = np.array([
    film_a,  # Film 1
    film_b,  # Film 2
    film_c   # Film 3
])

# 2. VEKTOR BOBOT (w) - Selera Hafidz
w = user

# 3. PREDIKSI (y) - Operasi Perkalian Matriks
# Ini yang disebut "Vectorization". Sekali pukul, kena semua.
y = X @ w 

print(f"Skor Semua Film: {y}")
# Output otomatis: [ 2 -2  0]

plt.bar_label(plt.bar(labels, y, color=colors))
plt.grid(axis="y", linestyle="--", alpha=0.3)
```
