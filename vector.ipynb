```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def plot_vectors(vecs: list, colors: list):
    plt.figure(figsize=(6, 6))
    plt.axvline(color="white", lw=1)
    plt.axhline(color="white", lw=1)
    
    origin = np.array([0,0])
    start = origin.copy()
    
    for i, vec in enumerate(vecs):
        plt.quiver(
            *start, *vec,
            angles='xy', scale_units='xy', scale=1,
            color=colors[i], label=f"v-{i+1}"
            )
        start += vec
    
    final = np.sum(vecs, axis=0)
    plt.quiver(
        *origin, *final,
        angles='xy', scale_units='xy', scale=1,
        color="purple", label=f"resultant ({final[0]},{final[1]})"
        )
    
    plt.grid(linestyle='--', alpha=0.5)
    
    limit = np.max(np.abs(np.cumsum(vecs, axis=0))) + 2
    plt.xlim(-limit, limit)
    plt.ylim(-limit, limit)
    plt.legend()
    plt.title("summation vector visualization")
    plt.style.use("dark_background")
    plt.show()
```

```python
a = np.array([3, 5])
b = np.array([-7, 4])
```

```python
plot_vectors([a, b], ["red", "green"])
```

```python
up = np.array([0,5])
down = np.array([0,-5])
right = np.array([5,0])
left = np.array([-5,0])

c_up = "red"
c_down = "green"
c_right = "blue"
c_left = "yellow"

gerak = [up, right, down, down, left, left, left, up, left, up]
colors = [c_up, c_right, c_down, c_down, c_left, c_left, c_left, c_up, c_left, c_up]

plot_vectors(gerak, colors)
```
