```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2 - 4*x + 6

def grad_f(x):
    return 2*x - 4
```

```python
x = -20
learning_rate = 0.1
history_x = [x]
history_y = [f(x)]

print(f"arah: {grad_f(x)}, posisi (x): {x}, ketinggian (y): {f(x)}")

while True:
    arah = grad_f(x)
    x -= learning_rate * arah
    y = f(x)
    history_x.append(x)
    history_y.append(y)
    print(f"arah: {arah}, posisi (x): {x}, ketinggian (y): {y}")
    
    if abs(arah) <= 0.0001:
        break
```

```python
plt.figure(figsize=(8, 8))
plt.axvline(color="grey", lw=1)
plt.axhline(color="grey", lw=1)
plt.grid(linestyle='--', alpha=0.3)

plt.plot(history_x, history_y, color="blue")
plt.scatter(history_x, history_y, c="red", )

plt.show()
```
