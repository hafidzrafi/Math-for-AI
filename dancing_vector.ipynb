### 🧠 KONSEP MATEMATIKA
Kita akan memutar vektor sebesar $\theta$ (theta) derajat berlawanan arah jarum jam.

**1. The Vector (Jarum Jam):**
Vektor awal $\vec{v}$ menunjuk ke arah Jam 3 (Sumbu X positif).
$$\vec{v} = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$$

**2. The Machine (Matriks Rotasi):**
[cite_start]Rumus sakti untuk memutar ruang sebesar sudut $\theta$[cite: 325]:
$$R = \begin{bmatrix} \cos(\theta) & -\sin(\theta) \\ \sin(\theta) & \cos(\theta) \end{bmatrix}$$
*(Catatan: NumPy menggunakan Radian, bukan Derajat. Jadi $45^\circ$ harus diubah jadi radian dulu).*

**3. The Transformation:**
[cite_start]Menggunakan perkalian matriks untuk mendapatkan posisi baru[cite: 326]:
$$\vec{v}_{baru} = R \cdot \vec{v}$$

---

### 📝 INSTRUKSI TUGAS

1.  **Setup Data:**
    * Buat vektor awal `v = [1, 0]`.
    * Tentukan sudut putar, misal `45` derajat. Konversi ke radian: `theta = np.radians(45)`.

2.  **Bangun Matriks:**
    * Buat array NumPy 2x2 sesuai rumus Matriks Rotasi di atas. Gunakan `np.cos(theta)` dan `np.sin(theta)`.

3.  **Eksekusi Rotasi:**
    * Lakukan perkalian matriks menggunakan operator `@` antara Matriks Rotasi dan Vektor `v`. Simpan hasilnya di `v_rotated`.

4.  **Visualisasi (Wajib):**
    * Gunakan `matplotlib`.
    * Plot vektor **ASLI** dengan warna **MERAH**.
    * Plot vektor **HASIL ROTASI** dengan warna **BIRU**.
    * Set limit sumbu x dan y dari -1.5 sampai 1.5 agar terlihat jelas.

> **Challenge:** Coba ganti sudutnya menjadi 90 atau 180 derajat dan lihat apakah jarumnya menunjuk ke arah yang benar!

```python
import numpy as np
import matplotlib.pyplot as plt
from main import vecs_visualize
```

```python
vec = np.array([10,0])
degree = 45
theta = np.radians(degree)

matrix_rotate = np.array([
    [np.cos(theta), np.sin(theta)],
    [-np.sin(theta), np.cos(theta)]
    ])

vec_rotated = vec @ matrix_rotate
```

```python
vecs_visualize([vec, vec_rotated], ["red", "blue"])
```
